```python
# Script to test functions

# packages
import numpy as np
import steepest_descent
```

```python
# implement function
def f(x):
    """
    Optimal point is (0.4694,−0.3878)
    """
    f_x = 10 * x[0]**2 + 2 * x[0] * x[1] + 5 * x[1]**2 - 8 * x[0] + 6 * x[1] + 20
    return f_x
```

```python
# test algorithm
starting_point = np.array([1, 1], dtype=np.float64)
optimal_point, func_eval, steps, diffs = steepest_descent.steepest_descent(
    starting_point, f, verbose=True
)
print(optimal_point)
print(func_eval)
print(steps)
print(diffs)
```
